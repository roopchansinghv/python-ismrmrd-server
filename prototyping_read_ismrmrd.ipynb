```python
for image_set_idx, image_set_unsorted in enumerate(image_sets):
   print("Image set %d" % image_set_idx)

   num_slices      = max(image_set_unsorted.headers[:]["slice"]) + 1
   num_repetitions = max(image_set_unsorted.headers[:]["repetition"]) + 1
   num_contrasts   = max(image_set_unsorted.headers[:]["contrast"]) + 1
   print (f"Max slice index, repetition, and contrast values in this image set are: {num_slices}, {num_repetitions}, and {num_contrasts}")

   image_matrix_0  = image_set_unsorted.headers[0]['matrix_size'][0]
   image_matrix_1  = image_set_unsorted.headers[0]['matrix_size'][1]

   read_dir        = image_set_unsorted.headers[0]['read_dir']
   phase_dir       = image_set_unsorted.headers[0]['phase_dir']
   slice_dir       = image_set_unsorted.headers[0]['slice_dir']
   print (f"Orientation vectors for this image are: {read_dir}, {phase_dir}, and {slice_dir}")

   for img_idx, each_img_header in enumerate(image_set_unsorted.headers):

      print (f"\n\n\nFor image {img_idx} in image set {image_set_idx}, headers are:")

      # for header_elements in each_img_header:

         # print (f"{header_elements}")

      for element_name in each_img_header.dtype.names:

         print (f"{element_name} : {each_img_header[element_name]}")
```
